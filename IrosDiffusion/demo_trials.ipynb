```python
import numpy as np
from irosbm.coords import angle2shift
from irosbm.optim import model_shadowgram
from irosbm.mask import codedmask


mpath: str = "/mnt/d/PhD_AASS/Coding/Images_fits/mask_NTHT_20260129_CORRECTED.fits"
wfm = codedmask(mpath, 2, 1)

rng = np.random.default_rng()
```

```python
poiss = rng.poisson(2, wfm.shape_detector)
```

```python
tx, ty = 42.0, 42.0   # [deg]
sx, sy = map(lambda theta: angle2shift(wfm, theta), (tx, ty))
sg = model_shadowgram(wfm, sx, sy, False, False, normalise=False)

det = poiss * sg
```

```python
poiss.mean(), poiss.std()
```

```python
det.mean(), det.std()
```

```python
mask = sg > 0

m = det[mask].sum() / mask.sum()
gamma = np.prod(wfm.shape_detector) / wfm.bulk.sum()

m, m * gamma
```
